```python

import numpy as np
input_text=open('/content/sample_data/input1.txt','r')
a=list(map(int,input_text.readline().split()))
n=int(a[0])
m=int(a[1])

entrypoint=list(map(int,input_text.readline().split()))
entry_row=int(entrypoint[0])
entry_column=int(entrypoint[1])

exitpoint=list(map(int,input_text.readline().split()))
exit_row=int(exitpoint[0])
exit_column=int(exitpoint[1])
current_pos=(entry_row,entry_column)

arr2D=np.zeros((n,m),dtype=str)
for i in range(n):
  line=input_text.readline()
  for j in range(m):
    arr2D[i][j]=line[j]

def Manhattan_distance(x1,y1,x2,y2):
  h_n= abs(x1-x2)+abs(y1-y2)
  return h_n

import heapq
p_queue=[]
visited=[]
ab=''
g_n=0
f_n=g_n+Manhattan_distance(current_pos[0],current_pos[1],exit_row,exit_column)
heapq.heappush(p_queue,(f_n,g_n,current_pos,ab))

while p_queue:
  f_n,g_n,(x,y),route = heapq.heappop(p_queue)

  if x==exit_row and y==exit_column:
    print(g_n)
    print(route)
    break

  if (x,y) in visited:
    continue
  visited.append((x,y))



  if x + 1 < n and arr2D[x+1][y] == '0' and (x+1,y) not in visited:
    new_g_n = g_n + 1
    new_f_n = new_g_n + Manhattan_distance(x+1, y, exit_row, exit_column)
    heapq.heappush(p_queue,(new_f_n,new_g_n,(x+1,y),route+'D'))

  if x - 1 >= 0 and arr2D[x-1][y] == '0' and (x-1,y) not in visited:
    new_g_n = g_n + 1
    new_f_n = new_g_n + Manhattan_distance(x-1, y, exit_row, exit_column)
    heapq.heappush(p_queue,(new_f_n,new_g_n,(x-1,y),route+'U'))

  if y + 1 < m and arr2D[x][y+1] == '0' and (x,y+1) not in visited:
    new_g_n = g_n + 1
    new_f_n = new_g_n + Manhattan_distance(x, y+1, exit_row, exit_column)
    heapq.heappush(p_queue,(new_f_n,new_g_n,(x,y+1),route+'R'))

  if y - 1 >= 0 and arr2D[x][y-1] == '0' and (x,y-1) not in visited:
    new_g_n = g_n + 1
    new_f_n = new_g_n + Manhattan_distance(x, y-1, exit_row, exit_column)
    heapq.heappush(p_queue,(new_f_n,new_g_n,(x,y-1),route+'L'))
else:
  print(-1)
```

```python
input_text=open('/content/sample_data/input2.txt','r')
n_m=list(map(int,input_text.readline().split()))
no_of_vertexes=int(n_m[0])
no_of_edges=int(n_m[1])

a_b=list(map(int,input_text.readline().split()))
initial=int(a_b[0])
goal=int(a_b[1])

heuristic_value=[]
for i in range(no_of_vertexes):
   a=list(map(int,input_text.readline().split()))
   vertex=int(a[0])
   h_value=int(a[1])
   t=(vertex,h_value)
   heuristic_value.append(t)


lst=[]
for j in range(no_of_edges):
   b=list(map(int,input_text.readline().split()))
   x=int(b[0])
   y=int(b[1])
   lst.append([x,y,1])

import heapq
###creating graph

graph = {}
for u,v,weight in lst:
    if u not in graph:
        graph[u] = []
    if v not in graph:
        graph[v] = []
    graph[u].append((v, weight))
    graph[v].append((u, weight))


###uniform cost search

def uniform_cost_search(graph, start, goal, no_of_vertexes):
    visited = [None] * (no_of_vertexes + 1)
    p_queue= [(0, start)]  # (cost, current_node)

    while p_queue:
        current_cost, current_node = heapq.heappop(p_queue)

        if current_node == goal:
            return current_cost
        if visited[current_node] is not None and current_cost > visited[current_node][0]:
            continue

        for neighbor,weight in graph[current_node]:
            total_cost = current_cost + weight
            if visited[neighbor] is None or total_cost < visited[neighbor][0]:

                visited[neighbor] = (total_cost, current_node)
                heapq.heappush(p_queue, (total_cost, neighbor))



    return None  # If goal unreachable


###asmissiblity_checking

inadmissible_nodes=[]
for k in range(no_of_vertexes):
  node=heuristic_value[k][0]
  actual_cost= uniform_cost_search(graph,node,goal,no_of_vertexes)
  if heuristic_value[k][1]>actual_cost:
    inadmissible_nodes.append(node)
  else:
    pass
if inadmissible_nodes==[]:
  print('1\nThe heuristic values are admissible')
else:
  print('0\nHere inadmissible nodes are',inadmissible_nodes)




```
