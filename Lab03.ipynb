```python
input2_text=open('/content/sample_data/input2.txt','r')
initial_nucleotides = input2_text.readline().strip().split(',')
target_gene_sequence = input2_text.readline().strip().split(',')
student_ID =list(map(int, input2_text.readline().strip().split()))
weight = student_ID[-4:]
booster = (student_ID[0] * 10 + student_ID[1])/ 100
default_weight = 1
alpha = -float('inf')
beta = float('inf')

def utility_calculation(current_sequence, target_gene_sequence, weight, booster_pos=None):
    value = 0
    new_weight=[]
    for i in weight:
        new_weight.append(i)

    if booster_pos is not None:
        for i in range(booster_pos, len(new_weight)):
            new_weight[i] *= booster

    for i in range(min(len(current_sequence), len(target_gene_sequence))):
        value += new_weight[i] * abs(ord(current_sequence[i]) - ord(target_gene_sequence[i]))

    for i in range(len(target_gene_sequence), len(current_sequence)):
        value += default_weight * abs(ord(current_sequence[i]) - 0)

    return -round(value, 2)

def algo(nucleotides, target_gene_sequence, alpha, beta, weight, current_seq=[], booster_pos=None,maximizing_player=True):
    if not nucleotides:
        return current_seq, utility_calculation(current_seq, target_gene_sequence, weight, booster_pos)

    if maximizing_player==True:
        best_sequence = []
        max_eval = -float('inf')
        for i in range(len(nucleotides)):
            nucleotide = nucleotides[i]
            new= nucleotides[:i] + nucleotides[i+1:]
            new_seq = current_seq + [nucleotide]
            new_booster_pos = booster_pos
            if nucleotide == 'S' and len(current_seq) == 0:
                new_booster_pos = len(new_seq) - 1
            sequence, value = algo(new, target_gene_sequence, alpha, beta, weight, new_seq,new_booster_pos,maximizing_player=False )
            if value > max_eval:
                max_eval = value
                best_sequence = sequence
            alpha = max(alpha, value)
            if alpha >= beta:
                break
        return best_sequence, max_eval
    else:
        min_eval = float('inf')
        best_sequence = []
        for i in range(len(nucleotides)):
            nucleotide = nucleotides[i]
            new = nucleotides[:i] + nucleotides[i+1:]
            new_seq = current_seq + [nucleotide]
            sequence, value = algo(new, target_gene_sequence, alpha, beta, weight, new_seq, booster_pos, maximizing_player=True)
            if value < min_eval:
                min_eval = value
                best_sequence = sequence
            beta = min(beta, value)
            if alpha >= beta:
                break
        return best_sequence, min_eval


best_gene1, utility_value1 = algo(initial_nucleotides, target_gene_sequence, alpha, beta, weight)
special_nucleotides = ['S'] + initial_nucleotides
best_gene2, utility_value2 = algo(special_nucleotides, target_gene_sequence, alpha, beta, weight)
if utility_value2 > utility_value1:
    print("\nYES")
    print("With special nucleotide")
    print("Best gene sequence generated: ", end="")
    for gene in best_gene2:
        print(gene, end="")
    print(",")
    print("Utility score:", utility_value2)
else:
    print("\nNO")
    print("Without special nucleotide")
    print("Best gene sequence generated: ", end="")
    for gene in best_gene1:
        print(gene, end="")
    print(",")
    print("Utility score:", utility_value1)



```
